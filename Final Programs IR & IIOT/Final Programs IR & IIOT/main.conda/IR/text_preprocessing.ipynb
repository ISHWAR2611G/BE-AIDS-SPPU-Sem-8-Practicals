### Assignment no 1: Write a program for pre-processing of a text document such as stop word removal, stemming

```python
import string
import re
import nltk
from nltk.corpus import stopwords
from nltk import WordPunctTokenizer 
from nltk.stem import WordNetLemmatizer, PorterStemmer
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
# nltk.download()
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('omw-1.4')
nltk.download('wordnet')
```

```python
tokenizer = WordPunctTokenizer()
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
```

```python
def text_preprocessing(text):
    text = text.lower()
    text = re.sub(r'\d+', '', text) # remove numbers
    text = re.sub(r'\s+', ' ', text) # remove extra spaces
    text = re.sub(r'[^\w\s]', '', text) # remove punctuation
    # print("Text after removing numbers, special characters, punctuations and white spaces:\n,", text)
    
    tokens = tokenizer.tokenize(text)
    # print("Tokens:\n", tokens)
    
    pos_tag = nltk.pos_tag(tokens)
    # print("Postags:\n", pos_tag)
    
    stop_words = set(stopwords.words('english'))
    tokens = [word for word in tokens if word not in stop_words]
    # print("Tokens after removing stopwords:\n", tokens)
    
    lemma_tokens = [lemmatizer.lemmatize(token) for token in tokens]
    # print("Lemmatized Tokens:\n", lemma_tokens, "\n")
    
    cleaned_text = ' '.join(lemma_tokens)
    return cleaned_text
    # print("Preprocessed text:\n", cleaned_text)
    
    
    
    
```
