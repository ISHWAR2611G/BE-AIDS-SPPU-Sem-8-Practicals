```python
# 1.  To apply the artificial immune pattern recognition to perform a task of structure damage Classification


import numpy as np

# --------------------------------
# Example Structural Data
# Feature1 = vibration
# Feature2 = strain
# 0 = No Damage
# 1 = Damage
# --------------------------------
X = np.array([
    [0.1, 0.2],
    [0.15, 0.25],
    [0.2, 0.3],
    [0.8, 0.9],
    [0.75, 0.85],
    [0.9, 0.95]
])

y = np.array([0, 0, 0, 1, 1, 1])

# --------------------------------
# AIS Model (Final Integrated)
# --------------------------------
class AIS:
    def __init__(self, num_detectors=4, mutation_rate=0.05):
        self.num_detectors = num_detectors
        self.mutation_rate = mutation_rate

    def train(self, X, y):
        # Step 1: Select detectors from data (learning)
        idx = np.random.choice(len(X), self.num_detectors, replace=False)
        self.detectors = X[idx]
        self.labels = y[idx]

        # Step 2: Mutation (AIS concept)
        mutation = np.random.normal(0, self.mutation_rate, self.detectors.shape)
        self.detectors = self.detectors + mutation

        print("Detectors:\n", self.detectors)

    def distance(self, a, b):
        return np.sqrt(np.sum((a - b) ** 2))

    def predict(self, sample):
        # Step 3: Find nearest detector (affinity)
        distances = [self.distance(sample, d) for d in self.detectors]
        nearest = np.argmin(distances)

        # Step 4: Return corresponding label
        return self.labels[nearest]

# --------------------------------
# Training
# --------------------------------
model = AIS()
model.train(X, y)

# --------------------------------
# Testing
# --------------------------------
print("\nClassification Results:")
correct = 0

for i in range(len(X)):
    pred = model.predict(X[i])
    print("Input:", X[i], "Predicted:", pred, "Actual:", y[i])

    if pred == y[i]:
        correct += 1

# Accuracy
accuracy = correct / len(X)
print("\nAccuracy:", accuracy)




# # code 2 : 
# import numpy as np

# # Sample Data
# X = np.array([
#     [0.1, 0.2],
#     [0.15, 0.25],
#     [0.2, 0.3],
#     [0.8, 0.9],
#     [0.75, 0.85],
#     [0.9, 0.95]
# ])

# y = np.array([0, 0, 0, 1, 1, 1])

# # Generate Detectors (from data instead of random)
# num_detectors = 4
# indices = np.random.choice(len(X), num_detectors, replace=False)
# detectors = X[indices]
# labels = y[indices]

# print("Detectors:\n", detectors)

# # Distance function
# def distance(a, b):
#     return np.sqrt(np.sum((a - b)**2))

# # Simple Prediction (nearest detector)
# def predict(sample):
#     distances = [distance(sample, d) for d in detectors]
#     nearest = np.argmin(distances)
#     return labels[nearest]

# # Testing
# print("\nResults:")
# correct = 0

# for i in range(len(X)):
#     pred = predict(X[i])
#     print("Input:", X[i], "Pred:", pred, "Actual:", y[i])
    
#     if pred == y[i]:
#         correct += 1

# print("\nAccuracy:", correct / len(X))
```
