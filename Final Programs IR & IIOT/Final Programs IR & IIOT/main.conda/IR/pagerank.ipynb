```python
import numpy as np

def pagerank(graph, damping_factor=0.85, max_iterations=100, tolerance=1e-6):
    # Step 1: Initialize variables
    num_nodes = len(graph)
    pagerank_values = {node: 1 / num_nodes for node in graph}
    new_pagerank_values = pagerank_values.copy()

    # Step 2: Compute the out-degree (number of outgoing links) for each node
    out_degree = {node: len(links) for node, links in graph.items()}


    # Step 3: Iteratively calculate PageRank values
    for iteration in range(max_iterations):
        for node in graph:
            # Step 4: Calculate PageRank based on incoming links
            rank_sum = 0
            for incoming_node in graph:
                if node in graph[incoming_node]:
                    rank_sum += pagerank_values[incoming_node] / out_degree[incoming_node]
            
            # Step 5: Apply the damping factor
            new_pagerank_values[node] = (1 - damping_factor) / num_nodes + damping_factor * rank_sum
        
        # Step 6: Check for convergence
        diff = sum(abs(new_pagerank_values[node] - pagerank_values[node]) for node in graph)
        pagerank_values = new_pagerank_values.copy()
        
        if diff < tolerance:
            print(f"Converged after {iteration + 1} iterations")
            break

    return pagerank_values

   


```

```python
 # Sample graph as described in the transcript (with nodes A, B, C, D)
graph = {
        'A': ['B'],
        'B': ['D'],
        'C': ['A', 'B', 'D'],
        'D': ['C']
}
    
pageranks = pagerank(graph)
print("PageRank Results:")
for node, rank in pageranks.items():
    print(f"Node {node}: {rank:.4f}")
```
