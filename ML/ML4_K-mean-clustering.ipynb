```python
#%% Import libraries
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
import warnings
warnings.filterwarnings('ignore')

#%% Load the dataset
df = pd.read_csv('Iris.csv')
print("Dataset Loaded Successfully!\n")
print(df.head())

#%% Prepare data for clustering
X = df.drop(['Id', 'Species'], axis=1)

#%% Scale the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

#%% Elbow Method to find the best k
wcss = []
for k in range(1, 11):
    model = KMeans(n_clusters=k, init='k-means++', random_state=42)
    model.fit(X_scaled)
    wcss.append(model.inertia_)

# Plot the Elbow curve
plt.figure(figsize=(8, 5))
plt.plot(range(1, 11), wcss, marker='o', linestyle='--', color='purple')
plt.title("Elbow Method - Find Optimal K")
plt.xlabel("Number of Clusters (k)")
plt.ylabel("WCSS")
plt.grid(True)
plt.show()

#%% Apply K-Means with k=3

kmeans = KMeans(n_clusters=3, init='k-means++', random_state=42)
df['Cluster'] = kmeans.fit_predict(X_scaled)

print("\nClustered Data (first 10 rows):")
print(df.head(10))

#%% Visualize Clustering Results
plt.figure(figsize=(12, 5))

# K-Means Cluster Plot
plt.subplot(1, 2, 1)
plt.scatter(X['PetalLengthCm'], X['PetalWidthCm'], c=df['Cluster'], cmap='viridis')
plt.title('K-Means Clusters (k=3)')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Petal Width (cm)')
plt.subplot(1, 2, 1)
plt.scatter(df['PetalLengthCm'], df['PetalWidthCm'], 
            c=df['Cluster'], cmap='viridis')
plt.title("K-Means Clusters (k=3)")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")

# Actual Species Plot
plt.subplot(1, 2, 2)
species_colors = df['Species'].astype('category').cat.codes
plt.scatter(df['PetalLengthCm'], df['PetalWidthCm'], 
            c=species_colors, cmap='viridis')
plt.title("Actual Species")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")

plt.tight_layout()
plt.show()

print("\n✅ The K-Means clusters closely match the actual species distribution!")
```

```python
# --- Import libraries ---
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

# --- Step 1: Load the dataset ---
df = pd.read_csv('Iris.csv')
print("First 5 rows of data:")
print(df.head())

# --- Step 2: Select features for clustering ---
# We remove 'Id' and 'Species' because:
# - 'Id' is just a number.
# - 'Species' is the real answer (we don't give it to the model).
X = df.drop(['Id', 'Species'], axis=1)

# --- Step 3: Scale all features ---
# Scaling helps K-Means work better (equal weight to all features).
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# --- Step 4: Find the best number of clusters (k) using Elbow Method ---
wcss = []  # WCSS = how spread out the clusters are
for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)  # Save the WCSS value

# --- Step 5: Plot the Elbow Curve ---
plt.plot(range(1, 11), wcss, marker='o')
plt.title("Elbow Method")
plt.xlabel("Number of Clusters (k)")
plt.ylabel("WCSS")
plt.show()

print("👉 The 'elbow' (bend) is around k = 3, so we choose 3 clusters.")

# --- Step 6: Apply K-Means with k=3 ---
kmeans = KMeans(n_clusters=3, random_state=42)
df['Cluster'] = kmeans.fit_predict(X_scaled)

print("\nCluster labels added to data:")
print(df.head())

# --- Step 7: Visualize the Clusters ---
plt.figure(figsize=(12,5))

# K-Means predicted clusters
plt.subplot(1,2,1)
plt.scatter(df['PetalLengthCm'], df['PetalWidthCm'], c=df['Cluster'], cmap='viridis')
plt.title("K-Means Clusters")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")

# Actual species
plt.subplot(1,2,2)
colors = df['Species'].astype('category').cat.codes
plt.scatter(df['PetalLengthCm'], df['PetalWidthCm'], c=colors, cmap='viridis')
plt.title("Actual Species")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")

plt.show()

print("\n✅ K-Means groups the flowers almost the same as the real species!")
```
