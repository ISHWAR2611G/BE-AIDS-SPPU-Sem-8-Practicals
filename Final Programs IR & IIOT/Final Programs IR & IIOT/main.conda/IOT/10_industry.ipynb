```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("IHMStefanini_industrial_safety_and_health_database_with_accidents_description.csv")
```

```python
df
```

```python
df = df.drop(columns=["Unnamed: 0"])
```

```python
df
```

```python
country_accident_count = df["Countries"].value_counts()
most = country_accident_count.idxmax(), country_accident_count.max()
least = country_accident_count.idxmin(), country_accident_count.min()
print("COUNTRY WITH MOST ACCIDENTS: ", most)
print("COUNTRY WITH LEAST ACCIDENTS: ", least)
```

```python
sector_accident_count = df["Industry Sector"].value_counts()
most = sector_accident_count.idxmax(), sector_accident_count.max()
least = sector_accident_count.idxmin(), sector_accident_count.min()
print("INDUSTRIAL SECTOR WITH MOST ACCIDENTS: ", most)
print("INDUSTRIAL SECTOR WITH LEAST ACCIDENTS: ", least)
```

```python
people_accident_count = df["Employee or Third Party"].value_counts()
most = people_accident_count.idxmax(), people_accident_count.max()
print("MOST LIKELY TO GET INTO ACCIDENT: ", most)
```

```python
gender_accident_count = df["Genre"].value_counts()
most = gender_accident_count.idxmax(), gender_accident_count.max()
print("GENDER MOST LIKELY TO GET INTO ACCIDENTS: ", most)
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=country_accident_count.index, y=country_accident_count.values)
plt.title("Accidents by Country")
plt.xlabel("Country")
plt.ylabel("Number of Accidents")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=sector_accident_count.index, y=sector_accident_count.values)
plt.title("Accidents by Industrial Sector")
plt.xlabel("Sector")
plt.ylabel("Number of Accidents")
plt.show()
```

```python
plt.figure(figsize=(6,6))
genderCount = df["Genre"].value_counts()
plt.pie(genderCount, labels=gender_accident_count.index, autopct="%1.1f%%")
plt.title("Accident by Gender")
plt.show()
```

```python
sns.countplot(x="Industry Sector", hue="Accident Level", data=df)
plt.show()
```
