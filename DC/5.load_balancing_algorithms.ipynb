```python
# ---------------------------------------------------------
# LOAD BALANCING SIMULATION (EXAM PERFECT VERSION)
# ---------------------------------------------------------

import random


# ---------------- SERVER CLASS ----------------

class Server:

    def __init__(self, server_id):
        self.server_id = server_id
        self.active_requests = 0


# ---------------- LOAD BALANCER ----------------

class LoadBalancer:

    def __init__(self, servers):
        self.servers = servers
        self.index = 0


    # Round Robin
    def round_robin(self):
        server = self.servers[self.index]
        self.index = (self.index + 1) % len(self.servers)
        return server


    # Random
    def random_choice(self):
        return random.choice(self.servers)


    # Least Connections
    def least_connections(self):
        return min(self.servers, key=lambda s: s.active_requests)


# ---------------- SIMULATION ----------------

def simulate(choice):

    servers = [Server(1), Server(2), Server(3)]
    lb = LoadBalancer(servers)

    print("\nLoad Balancing Started...")
    print("-" * 35)

    for i in range(1, 11):

        # Select algorithm
        if choice == 1:
            server = lb.round_robin()

        elif choice == 2:
            server = lb.random_choice()

        elif choice == 3:
            server = lb.least_connections()

        else:
            print("Invalid Choice")
            return

        # Assign request
        server.active_requests += 1

        print("Request", i, "→ Server", server.server_id)

    # Final Load
    print("\nFinal Server Loads:")

    for s in servers:
        print("Server", s.server_id, ":", s.active_requests)


# ---------------- MAIN ----------------

print("\nLOAD BALANCING SIMULATION")
print("1. Round Robin")
print("2. Random")
print("3. Least Connections")

choice = int(input("\nEnter choice (1/2/3): "))

simulate(choice)
```
