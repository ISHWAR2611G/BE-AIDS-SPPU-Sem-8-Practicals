```python
import requests
from bs4 import BeautifulSoup
```

```python
headers = {"user-agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/130.0.0.0 Safari/537.36"}
response = requests.get("https://www.free-powerpoint-templates-design.com/abstract-color-wave-powerpoint-templates/", headers = headers)
```

```python
soup = BeautifulSoup(response.text, 'html.parser')

div = soup.find_all('div', class_='download')
achor_tag = div[0].find_all('a')
link = achor_tag[0].get('href')

response = requests.get(link)
```

```python
import os

os.mkdir('ppt')
with open(f"ppt/1.pptx", 'wb') as f:
    f.write(response.content)
```

## video scrap

```python
url = "https://vimeo.com/"
header = {"user-agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/130.0.0.0 Safari/537.36"}
response = requests.get(url, header)
```

```python
soup = BeautifulSoup(response.text, 'html.parser')
videos = soup.find_all('video', class_ = 'w-full h-auto')
link = videos[0].get('src')
link
```

```python
import os
os.mkdir('video')
response = requests.get(link)

with open(f"video/video.mp4", 'wb') as f:
    f.write(response.content)
```
