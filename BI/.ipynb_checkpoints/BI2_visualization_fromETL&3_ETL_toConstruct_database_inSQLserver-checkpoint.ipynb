```python
# Step 1: SQL Code

# If using Microsoft SQL Server:

# -- Create Database
# CREATE DATABASE bi_lab;

# -- Use Database
# USE bi_lab;
# SHOW databases ;

# -- Create Sales Table
# CREATE TABLE sales_data (

#     order_number INT,

#     quantity_ordered INT,

#     price_each DECIMAL(10,2),

#     sales DECIMAL(10,2),

#     order_date VARCHAR(50),

#     status VARCHAR(30),

#     product_line VARCHAR(50),

#     country VARCHAR(50)
    
#     );



# -- Verify Table
# DESC sales_data;

# SELECT * FROM sales_data;

# DROP TABLE sales_data;
 
# DROP DATABASE bi_lab;







# Step 2: Python ETL Code
!pip install pandas sqlalchemy pymysql openpyxl


# ---------------------------------------------------------
# ETL PROCESS FOR SQL DATABASE
# ---------------------------------------------------------

import pandas as pd
from sqlalchemy import create_engine


# ---------------------------------------------------------
# Step 1 : Extract Data
# ---------------------------------------------------------


# For Excel use:
# data = pd.read_excel("sales_data.xlsx")

# For CSV use:
# data = pd.read_csv("sales_data_sample.csv")

data = pd.read_csv(
    "sales_data_sample.csv",
    encoding="latin1"
)

# or:

# data = pd.read_csv(
#     "sales_data_sample.csv",
#     encoding="cp1252"
# )


print("Source Data Loaded Successfully")
print(data.head())


# ---------------------------------------------------------
# Step 2 : Transform Data
# ---------------------------------------------------------

# Remove missing values
data = data.dropna()

# Remove duplicate values
data = data.drop_duplicates()

# Rename columns
data.columns = data.columns.str.lower()

print("\nData Transformation Completed")


# ---------------------------------------------------------
# Step 3 : Load Data into SQL Database
# ---------------------------------------------------------

engine = create_engine(
    "mysql+pymysql://root:ishwar2611%40G@localhost/bi_lab"
)

data.to_sql(
    name="sales_data",
    con=engine,
    if_exists="replace",
    index=False
)

print("\nData Loaded Successfully into SQL Database")
```
