```python
# ---------------------------------------------------------
# FUZZY SET OPERATIONS AND FUZZY RELATIONS
# ---------------------------------------------------------

def fuzzy_union(A, B):
    return {x: max(A[x], B[x]) for x in A}

def fuzzy_intersection(A, B):
    return {x: min(A[x], B[x]) for x in A}

def fuzzy_complement(A):
    return {x: round(1 - A[x], 2) for x in A}

def fuzzy_difference(A, B):
    return {x: min(A[x], 1 - B[x]) for x in A}

# ---------------------------------------------------------
# Cartesian Product
# ---------------------------------------------------------

def cartesian_product(A, B):
    return {(x, y): min(A[x], B[y]) for x in A for y in B}

# ---------------------------------------------------------
# Max-Min Composition
# ---------------------------------------------------------

def max_min_composition(R1, R2, X, Y, Z):
    return {
        (x, z): max(min(R1[(x, y)], R2[(y, z)]) for y in Y)
        for x in X for z in Z
    }

# ---------------------------------------------------------
# Display Function
# ---------------------------------------------------------

def display(title, data):
    print("\n" + title)
    for k, v in data.items():
        print(k, ":", v)

# ---------------------------------------------------------
# Main Program
# ---------------------------------------------------------

A = {'x1': 0.2, 'x2': 0.5, 'x3': 0.8}
B = {'x1': 0.6, 'x2': 0.4, 'x3': 0.3}

print("Fuzzy Set A =", A)
print("Fuzzy Set B =", B)

display("Union (A U B)", fuzzy_union(A, B))
display("Intersection (A ∩ B)", fuzzy_intersection(A, B))
display("Complement of A", fuzzy_complement(A))
display("Difference (A - B)", fuzzy_difference(A, B))

# ---------------------------------------------------------
# Fuzzy Relations
# ---------------------------------------------------------

C = {'y1': 0.3, 'y2': 0.7}
D = {'z1': 0.4, 'z2': 0.9}

print("\nFuzzy Set C =", C)
print("Fuzzy Set D =", D)

R1 = cartesian_product(A, C)
R2 = cartesian_product(C, D)

display("Relation R1 (A x C)", R1)
display("Relation R2 (C x D)", R2)

R = max_min_composition(R1, R2, A, C, D)
display("Max-Min Composition (R1 o R2)", R)
```
