```python
# ---------------------------------------------------------
# FUZZY SET OPERATIONS AND FUZZY RELATIONS
# ---------------------------------------------------------


# ---------------------------------------------------------
# Fuzzy Set Operations
# ---------------------------------------------------------

def fuzzy_union(A, B):

    result = {}

    for x in A:

        result[x] = max(A[x], B[x])

    return result



def fuzzy_intersection(A, B):

    result = {}

    for x in A:

        result[x] = min(A[x], B[x])

    return result



def fuzzy_complement(A):

    result = {}

    for x in A:

        result[x] = round(1 - A[x], 2)

    return result



def fuzzy_difference(A, B):

    result = {}

    for x in A:

        result[x] = min(A[x], 1 - B[x])

    return result



# ---------------------------------------------------------
# Cartesian Product
# ---------------------------------------------------------

def cartesian_product(A, B):

    relation = {}

    for x in A:

        for y in B:

            relation[(x, y)] = min(
                A[x],
                B[y]
            )

    return relation



# ---------------------------------------------------------
# Max-Min Composition
# ---------------------------------------------------------

def max_min_composition(R1, R2, X, Y, Z):

    relation = {}

    for x in X:

        for z in Z:

            values = []

            for y in Y:

                value = min(
                    R1[(x, y)],
                    R2[(y, z)]
                )

                values.append(value)

            relation[(x, z)] = max(values)

    return relation



# ---------------------------------------------------------
# Display Function
# ---------------------------------------------------------

def display(title, data):

    print("\n" + title)

    for key, value in data.items():

        print(key, ":", value)



# ---------------------------------------------------------
# Main Program
# ---------------------------------------------------------

A = {
    'x1': 0.2,
    'x2': 0.5,
    'x3': 0.8
}

B = {
    'x1': 0.6,
    'x2': 0.4,
    'x3': 0.3
}


print("Fuzzy Set A =", A)
print("Fuzzy Set B =", B)


# Set Operations

display(
    "Union (A U B)",
    fuzzy_union(A, B)
)

display(
    "Intersection (A ∩ B)",
    fuzzy_intersection(A, B)
)

display(
    "Complement of A",
    fuzzy_complement(A)
)

display(
    "Difference (A - B)",
    fuzzy_difference(A, B)
)



# ---------------------------------------------------------
# Fuzzy Relations
# ---------------------------------------------------------

C = {
    'y1': 0.3,
    'y2': 0.7
}

D = {
    'z1': 0.4,
    'z2': 0.9
}


print("\nFuzzy Set C =", C)
print("Fuzzy Set D =", D)


R1 = cartesian_product(A, C)

R2 = cartesian_product(C, D)


display(
    "Relation R1 (A x C)",
    R1
)

display(
    "Relation R2 (C x D)",
    R2
)



# Max-Min Composition

R = max_min_composition(
    R1,
    R2,
    A,
    C,
    D
)


display(
    "Max-Min Composition (R1 o R2)",
    R
)
```
