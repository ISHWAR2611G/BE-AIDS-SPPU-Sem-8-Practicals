```python
import re
from collections import defaultdict
```

```python
def preprocess(text):
    text = text.lower()
    text = re.sub(r'[^a-z0-9\s]', '', text)  # Remove punctuation
    tokens = text.split()
    return tokens
```

```python
def build_inverted_index(documents):
    inverted_index = defaultdict(list)
    
    for doc_id, text in enumerate(documents, start=1):
        words = preprocess(text)
        for word in set(words):  # Use set to avoid duplicates
            if doc_id not in inverted_index[word]:
                inverted_index[word].append(doc_id)
    
    return inverted_index  
```

```python
# Function to search for a query in the inverted index
def search_query(inverted_index, query):
    query_words = preprocess(query)
    if not query_words:
        return []
    
    result_docs = set(inverted_index.get(query_words[0], []))
    for word in query_words[2:]:
        result_docs &= set(inverted_index.get(word, []))
    
    return list(result_docs)
```

```python
# Sample documents
documents = [
    "The quick brown fox jumps over the lazy dog.",
    "Never jump over the lazy dog quickly.",
    "A brown dog is quick to jump.",
    "A lazy fox lies under the tree."
]

inverted_index = build_inverted_index(documents)
print("Inverted Index:")
dict(inverted_index)
```

```python
# Step 2: Query the inverted index
query = "quick brown"
result = search_query(inverted_index, query)

# Display results
print(f"\nQuery: '{query}'")
print(f"Documents containing the query: {result}")
for doc_id in result:
    print(f"Doc {doc_id}: {documents[doc_id-1]}")
```
